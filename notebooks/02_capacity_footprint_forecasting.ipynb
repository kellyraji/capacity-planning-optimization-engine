# 02 — Capacity Footprint Forecasting

## Portfolio objective

Convert the large server-hourly trace into a compact hourly capacity signal and generate a transparent 24-hour forecast.

### Engineering design

The raw Parquet partitions are streamed and aggregated immediately. The project does **not** concatenate the full raw server trace into one DataFrame.

### Forecasting signal

`Observed GPU Capacity(t) = sum of GPU count across observed servers at t`

The resulting `forecast_demand` is a **forecast observed GPU capacity footprint**, not measured utilization.


```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name.lower() == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from cpoe.config import SERVER_DATASET, PROCESSED
from cpoe.forecasting.pipeline import run_forecast_pipeline

print("Server dataset:", SERVER_DATASET)
print("Processed output:", PROCESSED)
```

```python
results = run_forecast_pipeline()

print("Hourly completeness:", results["completeness"])
display(results["metrics"])
display(results["summary"])
```

```python
hourly = results["hourly"]
forecast = results["forecast"]

print(f"Historical hours: {len(hourly):,}")
print(f"Historical peak GPU capacity: {hourly['observed_gpu_capacity'].max():,.0f}")
print(f"Historical mean GPU capacity: {hourly['observed_gpu_capacity'].mean():,.2f}")
print(f"Forecast peak GPU capacity: {forecast['forecast_demand'].max():,.0f}")

display(forecast)
```
